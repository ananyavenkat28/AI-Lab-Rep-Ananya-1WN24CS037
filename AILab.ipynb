Tic-Tac-Toe

```python
# Set up the game board initialized with the state shown in the image
board = ["O", "-", "X",
         "X", "-", "-",
         "X", "O", "O"]

# Initialize path cost to track the number of moves taken from this state
path_cost = 0

# Define a function to print the game board
def print_board():
    print(board[0] + " | " + board[1] + " | " + board[2])
    print(board[3] + " | " + board[4] + " | " + board[5])
    print(board[6] + " | " + board[7] + " | " + board[8])

# Define a function to handle a player's turn
def take_turn(player):
    global path_cost
    print(player + "'s turn.")
    position = input("Choose a position from 1-9: ")
    while position not in ["1", "2", "3", "4", "5", "6", "7", "8", "9"]:
        position = input("Invalid input. Choose a position from 1-9: ")
    position = int(position) - 1
    while board[position] != "-":
        position = int(input("Position already taken. Choose a different position: ")) - 1
    board[position] = player

    # Increment path cost for every valid move made
    path_cost += 1
    print_board()

# Define a function to check if the game is over
def check_game_over():
    # Check for a win
    if (board[0] == board[1] == board[2] != "-") or \
       (board[3] == board[4] == board[5] != "-") or \
       (board[6] == board[7] == board[8] != "-") or \
       (board[0] == board[3] == board[6] != "-") or \
       (board[1] == board[4] == board[7] != "-") or \
       (board[2] == board[5] == board[8] != "-") or \
       (board[0] == board[4] == board[8] != "-") or \
       (board[2] == board[4] == board[6] != "-"):
        return "win"
    # Check for a tie
    elif "-" not in board:
        return "tie"
    # Game is not over
    else:
        return "play"

# Define the main game loop
def play_game():
    print_board()

    # Automatically determine whose turn it is based on piece counts
    # The image has 3 'X's and 3 'O's, meaning it's X's turn next
    x_count = board.count("X")
    o_count = board.count("O")
    current_player = "X" if x_count <= o_count else "O"

    game_over = False
    while not game_over:
        take_turn(current_player)
        game_result = check_game_over()
        if game_result == "win":
            print(current_player + " wins!")
            print(f"Total path cost (moves made): {path_cost}")
            game_over = True
        elif game_result == "tie":
            print("It's a tie!")
            print(f"Total path cost (moves made): {path_cost}")
            game_over = True
        else:
            # Switch to the other player
            current_player = "O" if current_player == "X" else "X"

# Start the game
play_game()
```

Vaccum Cleaner Agent

```python
def vacuum_cleaner_agent(location, status):
    """Simple reflex agent for the vacuum cleaner problem."""
    print(f"Agent is located in Room: {location}")
    print(f"Room Status: {status}")

    # Action rules based on percepts
    if status[location] == "Dirty":
        return "Suck"
    elif location == "A":
        return "Move Right"
    elif location == "B":
        return "Move Left"
    else:
        return "NoOp"  # No operation if already clean and positioned


# Initialize environment state
environment = {"A": "Dirty", "B": "Dirty"}

# Initial agent location
agent_location = "A"

# Simulate 4 steps of the vacuum cleaner agent
for step in range(1, 5):
    print(f"\n--- Step {step} ---")
    action = vacuum_cleaner_agent(agent_location, environment)
    print(f"Action chosen: {action}")

    if action == "Suck":
        environment[agent_location] = "Clean"
        print(f"Room {agent_location} has been cleaned.")
    elif action == "Move Right":
        agent_location = "B"
        print("Moved to Room B.")
    elif action == "Move Left":
        agent_location = "A"
        print("Moved to Room A.")

    print(f"Updated Environment: {environment}")
```

8-Puzzle using DFS

```python
import heapq

GOAL_STATE = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

def find_position(state, value):
    for r in range(3):
        for c in range(3):
            if state[r][c] == value:
                return r, c

def manhattan_distance(state):
    distance = 0
    for r in range(3):
        for c in range(3):
            value = state[r][c]
            if value != 0:
                target_r, target_c = divmod(value - 1, 3)
                distance += abs(r - target_r) + abs(c - target_c)
    return distance

def get_neighbors(state):
    neighbors = []
    r, c = find_position(state, 0)

    moves = [(-1, 0, 'Up'), (1, 0, 'Down'), (0, -1, 'Left'), (0, 1, 'Right')]

    for dr, dc, direction in moves:
        new_r, new_c = r + dr, c + dc
        if 0 <= new_r < 3 and 0 <= new_c < 3:
            new_board = [list(row) for row in state]
            new_board[r][c], new_board[new_r][new_c] = new_board[new_r][new_c], new_board[r][c]
            neighbor_state = tuple(tuple(row) for row in new_board)
            neighbors.append((neighbor_state, direction))

    return neighbors

def solve_8_puzzle(start_state):
    start_h = manhattan_distance(start_state)
    open_set = [(start_h, 0, start_state, [])]

    visited = set()

    while open_set:
        f, g, current_state, path = heapq.heappop(open_set)

        if current_state == GOAL_STATE:
            return path

        if current_state in visited:
            continue
        visited.add(current_state)

        for neighbor, direction in get_neighbors(current_state):
            if neighbor not in visited:
                new_g = g + 1
                new_h = manhattan_distance(neighbor)
                new_f = new_g + new_h
                heapq.heappush(open_set, (new_f, new_g, neighbor, path + [direction]))

    return None

def print_state(state):
    for row in state:
        print(" ".join(str(tile) for tile in row))
    print()

if __name__ == "__main__":
    initial_puzzle = (
        (1, 2, 3),
        (0, 4, 6),
        (7, 5, 8)
    )

    print("Initial State:")
    print_state(initial_puzzle)

    solution_path = solve_8_puzzle(initial_puzzle)

    if solution_path is not None:
        print(f"🎉 Solution found in {len(solution_path)} moves!")
        print("Steps to solve:", " -> ".join(solution_path))
    else:
        print("❌ This puzzle configuration is unsolvable.")

print("Ananya 1WN24CS037")
```
